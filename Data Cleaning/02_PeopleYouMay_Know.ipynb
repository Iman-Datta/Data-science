```python
import json

def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
        return data
```

```python
def find_people_you_may_know(user_id, data):
    # Build user_id -> friends mapping
    user_friends = {}
    for user in data["users"]:
        user_friends[user["id"]] = set(user["friends"])

    # If user does not exist
    if user_id not in user_friends: # it check whether the user ID exists as a KEY in the dictionary(user_friends).
        return []

    direct_friends = user_friends[user_id]
    suggestions = {}

    # Look at friends of friends
    for friend in direct_friends:
        if friend not in user_friends:
            continue

        for mutual in user_friends[friend]:
            if mutual != user_id and mutual not in direct_friends:
                suggestions[mutual] = suggestions.get(mutual, 0) + 1

    # Sort by mutual-friend count (descending)
    sorted_suggestions = sorted(
        suggestions.items(),
        key=lambda x: x[1],
        reverse=True
    )

    # Return only user IDs
    return [uid for uid, _ in sorted_suggestions]
```

```python
data = load_data(r"D:\Data-science\Data clining\json\data.json")

try:
    user_id = int(input("Give the user id for which you are looking for recommendations: "))
except ValueError:
    print("Please enter a valid number")
else:
    recommendations = find_people_you_may_know(user_id, data)
    print("People you may know:", recommendations)
```

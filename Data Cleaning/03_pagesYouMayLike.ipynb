```python
import json

def load_page(filename):
    with open(filename, "r") as f:
        data = json.load(f)
        return data
```

```python
def pagesYouMayLike(user_id, data):
    user_friends = {}
    user_likeedPages = {}
    for user in data['users']:
        user_friends[user["id"]] = set(user['friends'])

    for user in data['users']:
        user_likeedPages[user["id"]] = set(user['liked_pages'])
    
    if user_id not in user_friends:
        return []
    
    direct_LikedPages = user_likeedPages[user_id]
    direct_friends = user_friends[user_id]
    suggestions = {}

    for friend in direct_friends:
        if friend not in user_likeedPages:
            continue
        liked_pages = user_likeedPages[friend]
        for mutual_pages in liked_pages:
            if mutual_pages not in direct_LikedPages:
                suggestions[mutual_pages] = suggestions.get(mutual_pages, 0) + 1
    sorted_suggestions = sorted(suggestions.items(), key = lambda x: x[1], reverse=True)
    return [page_id for page_id, _ in sorted_suggestions]
```

```python
data = load_page(r"D:\Data-science\Data clining\json\data.json")
uid = int(input("Give the user id for which you are looking for recommendations: "))
data = pagesYouMayLike(uid, data)
data
```

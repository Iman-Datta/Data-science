```python
import json

# Load the data from 
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
def display_users(data):
    print("Users and their connections:\n")

    for user in data['users']:
        print(
            f"User {user['name']} is friends with: {user['friends']} "
            f"and liked pages are: {user['liked_pages']}"
        )

    print("\nPages information")
    for page in data['pages']:
        print(f"ID{page['id']}: {page['name']}")
    return data
```

```python
data = load_data(r"D:\Anaconda project\data.json")
data = display_users(data)
print("\n Raw data (python dict: ): ")
data
```

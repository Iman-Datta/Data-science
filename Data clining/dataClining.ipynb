```python
import json

# it convert the json into python dict
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
def display_data(data):
    print("Users and their connections:\n")

    for user in data['users']:
        print(
            f"ID{user['id']}: User {user['name']} is friends with: {user['friends']} "
            f"and liked pages are: {user['liked_pages']}"
        )

    print("\nPages information")
    for page in data['pages']:
        print(f"ID{page['id']}: {page['name']}")
    return data
```

```python
def clean_data(data):
    # Remove users with missing / empty names
    data['users'] = [user for user in data['users']if user.get("name") not in (None, "")]

    # Remove pages with missing / empty names
    data['pages'] = [page for page in data['pages']if page.get("name") not in (None, "")]

    # Remove duplicate friends
    for user in data['users']:
        user['friends'] = list(set(user['friends']))

    # Remove duplicate liked pages
    for user in data['users']:
        user['liked_pages'] = list(set(user['liked_pages']))

    # Remove inactivate user
    data['users'] = [user for user in data['users'] if user['friends'] or user['liked_pages']]

    # new cleaned json file
    with open("cleaned_data.json", "w") as f:
        json.dump(data, f, indent=4)

    return data
```

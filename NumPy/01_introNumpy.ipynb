```python
import time

# Python list
size = 1_000
list1 = list(range(size))
list2 = list(range(size))

start = time.time()
result = [x + y for x, y in zip(list1, list2)]
end = time.time()

print(result)
print(end-start)
```

```python
import numpy as np
import time

# NumPy array
size = 1_000
array1 = np.array(list(range(size)))
array2 = np.array(list(range(size)))

start = time.time()
result = array1 + array2
end = time.time()

print(result)
print(end-start)
```

# 📘 NumPy Notes (Concise & Concept-Oriented)

## 1. What is NumPy?
NumPy (Numerical Python) is a core Python library used for **numerical computing**, especially in **Data Science, Machine Learning, and AI**.

It provides:
- Fast array operations
- Efficient memory usage
- Mathematical and statistical functions
- Support for multi-dimensional data

---

## 2. Why NumPy Instead of Python Lists?
Python lists are flexible but inefficient for numerical work.

### Limitations of Python Lists
- Store elements as pointers (not contiguous memory)
- Slow due to Python loops
- High memory overhead
- No built-in vectorized operations

### Advantages of NumPy
- Data stored in **contiguous memory**
- Implemented in **C (very fast)**
- Supports **vectorization** (no loops)
- Uses less memory
- Essential for ML & data analysis

**Key Idea:**  
> NumPy is faster because it avoids Python loops and runs operations in optimized C code. :contentReference[oaicite:0]{index=0}

---

## 3. NumPy Arrays (Core Concept)
A NumPy array:
- Is **homogeneous** (all elements same data type)
- Has a **fixed shape**
- Can be 1D, 2D, or n-D

### Important Properties
- `shape` → dimensions (rows, columns)
- `size` → total number of elements
- `ndim` → number of dimensions
- `dtype` → data type of elements

---

## 4. Memory Efficiency
NumPy arrays consume **much less memory** than Python lists because:
- Each element has a fixed size
- No extra object metadata per element

This is crucial when working with **large datasets** (lakhs or millions of values). :contentReference[oaicite:1]{index=1}

---

## 5. Vectorization (Very Important)
Vectorization means applying operations to **entire arrays at once**, instead of using loops.

### Why Vectorization Matters
- Loops in Python are slow
- NumPy executes operations in **parallel** using CPU optimizations (SIMD)
- Code becomes **shorter, cleaner, and faster**

**Rule to remember:**  
> If you are using loops with NumPy arrays, you are probably doing it wrong. :contentReference[oaicite:2]{index=2}

---

## 6. Creating Arrays Without Loops
NumPy provides shortcut functions to create arrays efficiently:
- Zeros array
- Ones array
- Identity matrix
- Range-based arrays
- Evenly spaced values

These avoid manual loops and improve performance.

---

## 7. Reshaping & Flattening
- **Reshaping** changes the structure of data without changing values
- **Flattening** converts multi-dimensional data into 1D

This is widely used in:
- Image processing
- ML model inputs
- Feature engineering

---

## 8. Indexing & Slicing
NumPy supports:
- Basic indexing (similar to Python lists)
- Slicing to extract subarrays

### Important Concept: Views vs Copies
- Slicing returns a **view**, not a copy
- Changes in sliced data affect the original array
- This improves performance and memory efficiency

Use `.copy()` if independent data is required. :contentReference[oaicite:3]{index=3}

---

## 9. Fancy Indexing & Boolean Masking
These techniques allow:
- Selecting multiple elements using index lists
- Filtering data using conditions

Very useful for:
- Data cleaning
- Feature selection
- Conditional analysis

---

## 10. Axes in NumPy
An **axis** represents a dimension.

- 1D → axis 0
- 2D → axis 0 (rows), axis 1 (columns)
- 3D → axis 0 (depth), axis 1 (rows), axis 2 (columns)

Operations along axes allow calculations **without loops**, such as row-wise or column-wise sums. :contentReference[oaicite:4]{index=4}

---

## 11. NumPy Data Types (dtype)
NumPy arrays are **strongly typed**.

Common dtypes:
- `int32`, `int64`
- `float32`, `float64`
- `bool`
- `complex`
- `object` (least efficient)

### Why dtype Matters
- Smaller dtype → less memory
- Smaller dtype → faster computation
- Wrong dtype → loss of precision

Choosing the right dtype is critical for performance optimization. :contentReference[oaicite:5]{index=5}

---

## 12. Broadcasting
Broadcasting allows NumPy to perform operations on arrays of **different shapes** without copying data.

### Key Rules
- Dimensions must be compatible
- Smaller array is “stretched” logically
- No extra memory is used

Used heavily in:
- Data normalization
- Feature scaling
- ML preprocessing

---

## 13. Built-in Mathematical Functions
NumPy provides optimized functions for:
- Mean, median, variance, standard deviation
- Min, max, sum, product
- Percentiles
- Cumulative operations
- Correlation
- Logarithmic and exponential operations

These functions are faster and more reliable than manual implementations. :contentReference[oaicite:6]{index=6}

---

## 14. Final Summary
- NumPy is the backbone of numerical computing in Python
- Arrays are faster, memory-efficient, and vectorized
- Avoid loops; rely on vectorization and broadcasting
- Understanding shape, axis, and dtype is essential
- NumPy is foundational for Pandas, ML, and Deep Learning

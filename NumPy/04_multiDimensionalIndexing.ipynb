```python
import numpy as np

arr = np.random.randint(10,50, size=(3,4))
print(arr)
```

```python
colSum = np.sum(arr, axis=0) # column-wise operation
print(colSum)
rowSum = np.sum(arr, axis=1) # row-wise operation
print(rowSum)
```

```python
val = arr[0,1] # Access the element at row index 0 and column index 1
print(val)
```

```python
val = arr[0:1]
print(val)
```

```python
# Slice rows from index 0 to 1 and columns from index 1 to 2 (end index excluded)

val = arr[0:2,1:3]
print(val)
```

```python
# Get all rows of the first column

first_col = arr[ : , 0]
print(first_col)
```

```python
# Get the first row each "sheet" in a 3D array

first_row = arr[0, :]
print(first_row)
```

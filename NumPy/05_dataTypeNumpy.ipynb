## 📌 NumPy Data Types (Short Note)

NumPy arrays are **homogeneous**, meaning all elements must have the **same data type (dtype)**.  
Choosing the correct dtype improves **memory efficiency** and **performance**.

---

### 🔹 Integer Types
- `int32` → 32-bit integer (uses less memory)
- `int64` → 64-bit integer (larger range, more memory)

👉 Use `int32` when values are small and memory matters.

---

### 🔹 Floating-Point Types
- `float32` → faster, less memory, lower precision
- `float64` → default, higher precision, more memory

👉 ML models often prefer `float32`.

---

### 🔹 Boolean Type
- `bool` → stores `True` / `False`
- Used in **masking and conditions**

---

### 🔹 Object Type
- `object` → can store mixed Python objects (lists, strings, dicts)
- ❌ Slow and memory-heavy
- ❌ Avoid for numerical computation

---

### 🔹 Why dtype Matters
- Smaller dtype → faster computation
- Smaller dtype → less memory usage
- Wrong dtype → precision loss or slow code

---

### 🔑 Key Takeaway
> Always choose the **smallest suitable dtype** for better performance.


```python
import numpy as np

arr = np.array([2,3,45,22,53,12])
datatype = arr.dtype
datatype
```

```python
arr = arr.astype("float64")
print(arr)
data_type = arr.dtype
print(data_type)
arr.nbytes
```

## 📌 Why Data Type Matters in NumPy

In NumPy, the **data type (dtype)** of an array directly affects **memory usage, performance, and precision**.

---

### 🔹 Memory Usage
- Smaller dtypes use less memory  
- Example: `int32` uses less memory than `int64`

---

### 🔹 Performance
- Operations on smaller dtypes are **faster**
- Less data is processed by the CPU

---

### 🔹 Precision
- Choosing the wrong dtype can **lose decimal accuracy**
- Example: `float32` has less precision than `float64`

---

### 🔹 Compatibility
- ML and GPU computations often require specific dtypes (e.g., `float32`)

---

### 🔑 Key Takeaway
> Choosing the correct dtype makes NumPy code **faster, memory-efficient, and accurate**.

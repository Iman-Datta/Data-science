```python
import numpy as np

arr = np.array([1,2,3,4,5,6,7])
newArr = arr ** 2 # Vectorized operation
print(newArr)
```

```python
newArr + 10
```

```python
arr1 = np.array([[1,2,3], [4,5,6]])
arr2 = np.array([7,8,9])

newArr = arr1 + arr2 # Broad Casting
print(newArr)
```

```python
data = np.random.randint(0, 100, size=(9,3))
mean = data.mean(axis=0)
std = data.std(axis=0)

normalized_data = (data - mean) / std # Formula: (value − mean) / standard deviation
print(normalized_data)
```

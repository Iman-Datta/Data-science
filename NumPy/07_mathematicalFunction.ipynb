```python
import numpy as np


arr = np.random.randint(0, 100, size=(9,3))
print(f"Original matrix: \n {arr}")
```

```python
mean = np.mean(arr)
print(mean)
```

```python
min = np.min(arr)
print(min)
```

```python
max = np.max(arr)
print(max)
```

```python
std = np.std(arr)
print(std)
```

```python
avg = np.median(arr)
print(avg)
```

```python
sum = np.sum(arr)
product = np.prod(arr)
print(f"The sum of the matrix is: {sum}, and the product of the matrix is: {product}")
```

```python
percentile = np.percentile(arr, 60)
print(percentile)
```

```python
unique = np.unique(arr)
print(unique)
```

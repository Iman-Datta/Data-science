```python
import numpy as np

# 1D array:
array = np.array([1,2,3,4,5])
print(type(array))
array
```

```python
# 2D array:
array = np.array([[1,2,3,4],[5,6,7,8]]) 
print(type(array))
print(array)
```

```python
# Creating a 2D NumPy array filled with zeros (3 rows × 4 columns)

np.zeros ((3,4))
```

```python
# Creating a 2D NumPy array filled with ones (3 rows × 4 columns)

np.ones((3,3))
```

```python
# Create a 2×3 NumPy array where all elements are filled with 4

np.full((2,3), 4)
```

```python
# Indentity Matrix

np.eye(5)
```

```python
# Create an array starting from 0 up to 100 (excluding 100) with a step of 10

np.arange(0,100,10)
```

```python
# Evenly Speced aray

np.linspace(1,100,5)
```

```python
import numpy as np

# Create a 3×5 NumPy matrix with random integers between 37 and 99
matrix = np.random.randint(37, 100, size=(3, 5))

# Print the matrix
print(f"Matrix: \n{matrix}")


print(f"the shape of the matrix: {matrix.shape}")   # Print the shape of the matrix (rows, columns)
print(f"the total number of elements: {matrix.size}")    # Print the total number of elements in the matrix
print(f"the total number of elements: {matrix.ndim}")    # Print the number of dimensions of the matrix
```

```python
# Reshape the matrix from 3×5 to 5×3 (total elements remain the same)

reshapeMatrix = matrix.reshape((5,3))
print(reshapeMatrix)
```

```python
# Convert 2D array to 1D array

print(matrix.flatten())
```

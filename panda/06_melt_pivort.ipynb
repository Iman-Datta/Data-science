```python
import pandas as pd

df = pd.read_csv(r"D:\Data-science\panda\File\06_melt_pivort.csv")
df
```

```python
df2 = df.melt(id_vars=["Name"], #  Columns to keep fixed
            value_vars=["Math", "Science", "English"], # Columns to unpivot
            var_name="Subject", # Name of the new column for subjects
            value_name="Score").copy().sort_values(["Name"]) # Name of the new column for values

df2
```

```python
df2.pivot(index=["Name"], columns=["Subject"], values=["Score"]) # Convert long format back to wide format (subjects as columns)
```
